# Challenge 2 — List of Dictionaries

```python
trips = [
    {"zone": "Downtown", "wait": 4.5, "fare": 18},
    {"zone": "Airport", "wait": 12.1, "fare": 42},
    {"zone": "Downtown", "wait": 7.8, "fare": 24},
    {"zone": "Airport", "wait": 15.2, "fare": 51}
]
```

```python
# How many trips started at the Airport?

airport_count = 0

for trip in trips:
    if trip["zone"] == "Airport":
        airport_count += 1

print("Airport trips:", airport_count)
```

```python
# Average fare

total_fare = 0

for trip in trips:
    total_fare += trip["fare"]

average_fare = total_fare / len(trips)

print("Average fare:", average_fare)
```

```python
# How many trips had wait > 10?

long_wait_count = 0

for trip in trips:
    if trip["wait"] > 10:
        long_wait_count += 1

print("Trips with long waits:", long_wait_count)
```

```python
# Highest fare

highest_fare = trips[0]["fare"]

for trip in trips:
    if trip["fare"] > highest_fare:
        highest_fare = trip["fare"]

print("Highest fare:", highest_fare)
```

# Challenge 3 — Service-Level Function

# Challenge 4 — Data-Quality Check

```python
wait_times = [5.2, 6.3, -3, 8.4, 500, 7.1]
```

```python
for x in wait_times:

    if x < 0:
        print(x, "Invalid: negative waiting time")

    elif x > 120:
        print(x, "Suspicious: extremely large waiting time")

    else:
        print(x, "Looks reasonable")
```

```python
suspicious_values = []

for x in wait_times:

    if (x < 0) or (x > 120):
        suspicious_values.append(x)

print(suspicious_values)
```

# Final Mini-Challenge

```python
trips = [
    {"zone": "Downtown", "wait": 5.2, "fare": 18.5},
    {"zone": "Airport", "wait": 13.4, "fare": 45.2},
    {"zone": "West", "wait": 8.7, "fare": 23.8},
    {"zone": "Airport", "wait": 16.1, "fare": 52.0},
    {"zone": "Downtown", "wait": 4.6, "fare": 17.2}
]


# --------------------------------------------------
# 1. Number of trips
# --------------------------------------------------

number_of_trips = len(trips)

print("Number of trips:", number_of_trips)


# --------------------------------------------------
# 2. Average fare
# --------------------------------------------------

total_fare = 0

for trip in trips:
    total_fare += trip["fare"]

average_fare = total_fare / len(trips)

print(f"Average fare: ${average_fare:.2f}")


# --------------------------------------------------
# 3. Average waiting time
# --------------------------------------------------

total_wait = 0

for trip in trips:
    total_wait += trip["wait"]

average_wait = total_wait / len(trips)

print(f"Average waiting time: {average_wait:.1f} minutes")


# --------------------------------------------------
# 4. Number of trips with wait > 10
# --------------------------------------------------

long_wait_count = 0

for trip in trips:

    if trip["wait"] > 10:
        long_wait_count += 1

print("Trips with wait > 10:", long_wait_count)


# --------------------------------------------------
# 5. Percentage with wait > 10
# --------------------------------------------------

percentage_long = long_wait_count / len(trips) * 100

print(f"Percentage with long wait: {percentage_long:.1f}%")


# --------------------------------------------------
# 6. Count trips by zone
# --------------------------------------------------

zone_counts = {}

for trip in trips:

    zone = trip["zone"]

    if zone in zone_counts:
        zone_counts[zone] += 1
    else:
        zone_counts[zone] = 1

print("Trips by zone:", zone_counts)


# --------------------------------------------------
# 7. Service-level classification
# --------------------------------------------------

def service_level(wait):

    if wait <= 8:
        return "Good"

    elif wait <= 12:
        return "Warning"

    else:
        return "Poor"


print("\nService levels:")

for trip in trips:

    level = service_level(trip["wait"])

    print(
        trip["zone"],
        trip["wait"],
        level
    )


# --------------------------------------------------
# 8. Airport trips
# --------------------------------------------------

print("\nAirport trips:")

for trip in trips:

    if trip["zone"] == "Airport":
        print(trip)


# --------------------------------------------------
# 9. Highest fare
# --------------------------------------------------

highest_fare_trip = trips[0]

for trip in trips:

    if trip["fare"] > highest_fare_trip["fare"]:
        highest_fare_trip = trip

print("\nHighest-fare trip:")
print(highest_fare_trip)
```
