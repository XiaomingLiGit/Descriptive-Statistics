# Part B — Lists and Loops

```python
wait_times = [
    4.5, 7.2, 11.3, 6.8, 14.6,
    9.1, 5.4, 16.2, 8.0, 12.5
]
```

```python
# Question 6 — Basic List Operations

print(len(wait_times))
print(min(wait_times))
print(max(wait_times))
print(sum(wait_times))
```

```python
# Question 7 — Mean Waiting Time

mean_wait = sum(wait_times) / len(wait_times)

print(mean_wait)
```

```python
# Question 8 — Count Long Waits

long_wait_count = 0

for wait in wait_times:
    if wait > 10:
        long_wait_count += 1

print(long_wait_count)
```

```python
# Question 9 — Percentage of Long Waits

percentage_long_wait = (
    long_wait_count / len(wait_times)
) * 100

print(percentage_long_wait)
```

```python
# Question 10 — Create a Filtered List

long_waits = []

for wait in wait_times:
    if wait > 10:
        long_waits.append(wait)

print(long_waits)
```

```python
# Question 11 — Data Quality

wait_times_check = [
    5.2, 7.1, -4.0, 8.4, 150.0, 6.9
]

for wait in wait_times_check:
    if wait < 0 or wait > 120:
        print("Suspicious value:", wait)
```

```python
# Question 13 — Fare per Kilometre Function

def fare_per_km(fare, distance):
    return fare / distance


print(fare_per_km(30, 10))
```

```python
# Question 14 — Waiting-Time Category

def wait_category(wait):
    if wait < 5:
        return "Short"
    elif wait <= 10:
        return "Medium"
    else:
        return "Long"
    
    
print(wait_category(4))
print(wait_category(7))
print(wait_category(12))
```

```python
# Question 15 — Service-Level Function

def service_level(wait):
    if wait <= 8:
        return "Good"
    elif wait <= 12:
        return "Warning"
    else:
        return "Poor"
    
wait_times = [4.5, 8.7, 12.4, 6.2, 15.1]

for wait in wait_times:
    print(wait, service_level(wait))
```

# Part D — Lambda and NumPy

```python
# Question 16 — Lambda Expression

def double(x):
    return x * 2

double = lambda x: x * 2

print(double(6))
```

```python
# Question 17 — Classification with Lambda

wait_label = lambda wait: (
    "Long" if wait > 10 else "Normal"
)


print(wait_label(6))
print(wait_label(15))
```

```python
# Question 18 — NumPy Array

import numpy as np

x = np.array([4, 6, 8, 10, 12])

print(np.mean(x))
print(np.std(x))
print(np.min(x))
print(np.max(x))
```

```python
# Question 19 — Vectorized Calculation

x_scaled = x * 1.5

print(x_scaled)
```

Question 20 — Short Explanation

A NumPy expression such as x * 1.5 applies the operation to every element of the array at once. It is shorter and easier to read than manually writing a loop and is generally more efficient for numerical calculations.

# Part E — Mini Data-Analysis Case

```python
trips = [
    {"zone": "Downtown",   "wait": 4.8,  "fare": 19.5},
    {"zone": "Airport",    "wait": 14.2, "fare": 47.0},
    {"zone": "West",       "wait": 8.9,  "fare": 25.5},
    {"zone": "Airport",    "wait": 12.8, "fare": 43.5},
    {"zone": "Downtown",   "wait": 6.1,  "fare": 21.0},
    {"zone": "University", "wait": 7.4,  "fare": 18.5},
    {"zone": "Airport",    "wait": 16.5, "fare": 55.0},
    {"zone": "West",       "wait": 9.6,  "fare": 27.0}
]
```

```python
# Question 21 — Number of Trips

total_trips = len(trips)

print(total_trips)
```

```python
# Question 22 — Average Fare

total_fare = 0

for trip in trips:
    total_fare += trip["fare"]

average_fare = total_fare / len(trips)

print(average_fare)
```

```python
# Question 23 — Average Waiting Time

total_wait = 0

for trip in trips:
    total_wait += trip["wait"]

average_wait = total_wait / len(trips)

print(average_wait)
```

```python
# Question 24 — Long Waits


long_wait_count = 0

for trip in trips:
    if trip["wait"] > 10:
        long_wait_count += 1

long_wait_percentage = (
    long_wait_count / len(trips)
) * 100

print(long_wait_count)
print(long_wait_percentage)
```

```python
# Question 25 — Airport Trips

for trip in trips:
    if trip["zone"] == "Airport":
        print(trip)
```

```python
# Question 26 — Highest-Fare Trip

highest_fare_trip = trips[0]

for trip in trips:
    if trip["fare"] > highest_fare_trip["fare"]:
        highest_fare_trip = trip

print(highest_fare_trip)
```

```python
# Question 27 — Count Trips by Zone

zone_counts = {}

for trip in trips:
    zone = trip["zone"]

    if zone in zone_counts:
        zone_counts[zone] += 1
    else:
        zone_counts[zone] = 1

print(zone_counts)
```

```python
# Question 28 — Service Classification

def service_level(wait):
    if wait <= 8:
        return "Good"
    elif wait <= 12:
        return "Warning"
    else:
        return "Poor"
    
    
for trip in trips:
    print(
        trip["zone"],
        trip["wait"],
        service_level(trip["wait"])
    )
```
