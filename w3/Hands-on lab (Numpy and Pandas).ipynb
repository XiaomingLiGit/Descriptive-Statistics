# Challenge 1 — Analyze Waiting Times



```python
import numpy as np
```

```python
wait_times = [3.8, 7.1, 12.5, 6.3, 15.2, 4.9, 11.0]
```

```python
#  1. Calculate the mean

mean_wait = sum(wait_times) / len(wait_times)

print(mean_wait)
```

```python
wait_times = np.array(wait_times)

np.mean(wait_times)
```

```python
print(round(mean_wait, 3))
```

```python
# 2. Calculate min/max

print("Minimum:", min(wait_times))
print("Maximum:", max(wait_times))

np.max(wait_times)
np.min(wait_times)
```

```python
# 3. Count waits > 10

long_wait_count = 0

for wait in wait_times:
    if wait > 10:
        long_wait_count += 1

print(long_wait_count)
```

```python
# 4. Classify each wait

def wait_category(wait):
    if wait < 5:
        return "Short"
    elif wait < 10:
        return "Medium"
    else:
        return "Long"
    
    
for wait in wait_times:
    print(wait, wait_category(wait))
```

```python
# 5. Percentage of long waits

percentage_long = long_wait_count / len(wait_times) * 100

print(percentage_long)
```

```python
print(f"{percentage_long:.2f}%")
```

# Challenge 3 — Service Rule

```python
Good:     wait <= 8
Warning:  8 < wait <= 12
Poor:     wait > 12
```

```python
def service_level(wait):
    if wait <= 8:
        return "Good"
    elif wait <= 12:
        return "Warning"
    else:
        return "Poor"
```

```python
print(service_level(5))
print(service_level(10))
print(service_level(15))
```

```python
wait_times = [4.2, 8.5, 6.1, 12.3, 15.2]

for wait in wait_times:
    print(wait, service_level(wait))
```

# Challenge 4 — Data-Quality Thinking

```python
wait_times = [5.2, 6.3, -3, 8.4, 500, 7.1]
```

```python
Question:
Which values look suspicious?    # outlier / anomoly detection, fraud detection... (context)

Reason:
- -3 is impossible because waiting time cannot normally be negative.
- 500 minutes is extremely large and should be investigated.
```

```python
for wait in wait_times:
    if wait < 0:
        print(wait, "Invalid: negative waiting time")
    elif wait > 120:
        print(wait, "Suspicious: unusually large waiting time")
```

```python
suspicious_values = []

for wait in wait_times:
    if wait < 0 or wait > 120:
        suspicious_values.append(wait)

print(suspicious_values)
```

# Final Mini-Challenge

```python
trips = [
    {"zone": "Downtown", "wait": 5.2, "fare": 18.5},
    {"zone": "Airport", "wait": 13.4, "fare": 45.2},
    {"zone": "West", "wait": 8.7, "fare": 23.8},
    {"zone": "Airport", "wait": 16.1, "fare": 52.0},
    {"zone": "Downtown", "wait": 4.6, "fare": 17.2}
]
```

```python
# 1. How many trips?

len(trips)
```

```python
# 2. Average fare

total_fare = 0

for trip in trips:
    total_fare += trip["fare"]

average_fare = total_fare / len(trips)

print(average_fare)
```

```python
# 3. Average waiting time

total_wait = 0

for trip in trips:
    total_wait += trip["wait"]

average_wait = total_wait / len(trips)

print(average_wait)
```

```python
# 4. How many trips have wait > 10?

long_wait_count = 0

for trip in trips:
    if trip["wait"] > 10:
        long_wait_count += 1

print(long_wait_count)
```

```python
# 5. Percentage of trips with wait > 10

percentage = long_wait_count / len(trips) * 100

print(percentage)
```

```python
# 6. Which zone appears most often?


zone_counts = {}

for trip in trips:
    zone = trip["zone"]

    if zone in zone_counts:
        zone_counts[zone] += 1
    else:
        zone_counts[zone] = 1

print(zone_counts)
```

```python
# 7. Classify every trip as Good / Warning / Poor

def service_level(wait):
    if wait <= 8:
        return "Good"
    elif wait <= 12:
        return "Warning"
    else:
        return "Poor"
    
    
for trip in trips:
    print(
        trip["zone"],
        trip["wait"],
        service_level(trip["wait"])
    )    
```

```python
# 8. Print only Airport trips

for trip in trips:
    if trip["zone"] == "Airport":
        print(trip)
```

```python
# 9. Find the highest fare

highest_fare = trips[0]["fare"]

for trip in trips:
    if trip["fare"] > highest_fare:
        highest_fare = trip["fare"]

print(highest_fare)
```

```python
# trip

highest_fare_trip = trips[0]

for trip in trips:
    if trip["fare"] > highest_fare_trip["fare"]:
        highest_fare_trip = trip

print(highest_fare_trip)
```

10. Two-sentence interpretation

The average waiting time was 9.6 minutes, and 40% of trips had waiting times longer than 10 minutes. 
Both poor-service observations occurred at the Airport, suggesting that Airport waiting times may deserve further investigation.

```python
# 4. Mean, median, and SD of waiting time

df["waiting_time_min"]
```

```python
mean_wait = df["waiting_time_min"].mean()
median_wait = df["waiting_time_min"].median()
sd_wait = df["waiting_time_min"].std()

print(f"Mean: {mean_wait:.2f}")
print(f"Median: {median_wait:.2f}")
print(f"SD: {sd_wait:.2f}")
```

```python
# 5. Median waiting time by zone

df.groupby("zone")["waiting_time_min"].median()
```

```python
# 6. Filter trips with waiting time > 10

long_waits = df[df["waiting_time_min"] > 10]

long_waits

# if cond_1 & cond_2:
```

```python
# 7. Plot a histogram

df["waiting_time_min"].plot.hist(
    bins=15,
    edgecolor="black"
)

plt.xlabel("Waiting Time (minutes)")
plt.ylabel("Frequency")
plt.title("Distribution of Passenger Waiting Time")
plt.show()
```

The median waiting time is 4.8 minutes, while the mean is slightly higher at 5.18 minutes, 
suggesting that the distribution is somewhat right-skewed. Most trips have relatively short waiting times, 
but a few unusually long waits, including values above 20 and 30 minutes, increase the variability and pull the mean upward.
