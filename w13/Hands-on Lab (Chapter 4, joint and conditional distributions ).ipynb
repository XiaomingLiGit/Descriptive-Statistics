```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df = pd.read_csv('trips.csv')
```

```python
df["late"] = df["waiting_time_min"] > 8
df["peak"] = df["hour"].isin([7,8,9,16,17,18])

pd.crosstab(df["late"], df["peak"])
pd.crosstab(df["late"], df["peak"], normalize=True)
pd.crosstab(df["late"], df["peak"], normalize="columns")
```

```python
# Joint distribution: all cells sum to 1
joint = pd.crosstab(df["late"], df["peak"], normalize=True)

# Conditional distribution P(late | peak)
cond = pd.crosstab(df["late"], df["peak"], normalize="columns")

# Marginal distribution
late_prob = df["late"].value_counts(normalize=True)
```
