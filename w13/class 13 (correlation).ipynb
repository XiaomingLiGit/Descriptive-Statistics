# Distributions

```python
from scipy import stats

# Discrete distributions
stats.bernoulli.pmf(1, p=0.35)
stats.binom.pmf(k=4, n=10, p=0.35)
stats.poisson.pmf(k=3, mu=4)

```

```python
# Continuous distributions
stats.norm.pdf(x=0, loc=0, scale=1)
stats.expon.pdf(x=2, scale=1/0.8)


```

```python
# CDF and quantiles
stats.norm.cdf(1.96)
stats.norm.ppf(0.975)

```

```python
# Random samples
stats.poisson.rvs(mu=4, size=1000, random_state=42)
```

# Compute probability using SciPy

```python
from scipy import stats

# Binomial: P(X = 4), n=5, p=0.8
stats.binom.pmf(4, n=5, p=0.8)

# Normal: P(X <= 12), mean=10, sd=2
stats.norm.cdf(12, loc=10, scale=2)

# Poisson: P(X >= 5), lambda=3
1 - stats.poisson.cdf(4, mu=3)
```

```python
import numpy as np

rng = np.random.default_rng(42)
rolls = rng.integers(1, 7, size=10_000)

# empirical probability of rolling a 6
(rolls == 6).mean()

# theoretical probability
1 / 6
```
