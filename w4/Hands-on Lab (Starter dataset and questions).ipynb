```python
# Full Version

# Make a copy
clean_df = df.copy()

# 1. Remove exact duplicates
clean_df = clean_df.drop_duplicates()

# 2. Standardize mode labels
clean_df['mode'] = clean_df['mode'].str.title()

# 3. Replace impossible distance with missing
clean_df.loc[
    clean_df['distance_km'] < 0,
    'distance_km'
] = None

# 4. Treat 999 as suspicious / missing
clean_df.loc[
    clean_df['wait_min'] == 999,
    'wait_min'
] = None

clean_df
```
