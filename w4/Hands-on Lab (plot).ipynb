```python
Lab 1 — Line Plot: Trips by Hour
    
Tasks:
1. Create a line plot of hours versus trips.
2. Use circle markers for the data points.
3. Add the title: Trips by Hour.
4. Label the x-axis as Hour.
5. Label the y-axis as Number of Trips.
6. Add grid lines.
7. Based on the graph, identify the hour with the highest number of trips.
```

```python
hours = np.array([6, 8, 10, 12, 14, 16, 18, 20])
trips = np.array([35, 72, 58, 65, 70, 110, 125, 68])

plt.plot(hours, trips, marker='o')

plt.title("Trips by Hour")
plt.xlabel("Hour")
plt.ylabel("Number of Trips")

plt.grid()

plt.show()

# The highest number of trips occurs at 18:00, with 125 trips.
```

```python
Lab 2 — Compare Two Days

Tasks:
1. Plot weekday and weekend trip volumes on the same graph.
2. Use different markers for the two lines.
3. Add a legend.
4. Add an appropriate title and axis labels.
5. Add grid lines.
6. At which hour is the difference between weekday and weekend trip volume the largest?
7. Write one sentence describing the main pattern you observe.
```

```python
hours = np.array([6, 9, 12, 15, 18, 21])

weekday = np.array([30, 85, 62, 70, 120, 55])
weekend = np.array([20, 45, 68, 80, 95, 78])

plt.plot(hours, weekday, marker='o', label='Weekday')
plt.plot(hours, weekend, marker='s', label='Weekend')

plt.title("Trips by Hour: Weekday vs Weekend")
plt.xlabel("Hour")
plt.ylabel("Number of Trips")

plt.grid()
plt.legend()

plt.show()

# The largest difference occurs around 9:00, when weekday demand is much higher than weekend demand.
```

```python
# Lab 3 — Scatterplot: Distance vs Fare


Tasks:
1. Create a scatterplot of distance versus fare.
2. Label the x-axis as Distance (km).
3. Label the y-axis as Fare ($).
4. Add the title: Trip Distance vs Fare.
5. Add grid lines.
6. Describe the direction and strength of the relationship.
7. Can you conclude that longer distance causes higher fare based only on this graph? Explain briefly.
```

```python
distance = np.array([2, 3, 5, 7, 8, 10, 12, 15, 18, 20])
fare = np.array([9, 11, 15, 20, 22, 27, 31, 38, 45, 49])

plt.scatter(distance, fare)

plt.title("Trip Distance vs Fare")
plt.xlabel("Distance (km)")
plt.ylabel("Fare ($)")

plt.grid()

plt.show()

# There is a strong positive relationship between trip distance and fare. Longer trips generally have higher fares.
```

```python
# Lab 4 — Scatterplot with a Third Variable

Tasks:
1. Create a scatterplot of distance versus fare.
2. Use marker size to represent wait.
3. Add an appropriate title and axis labels.
4. Use transparency (alpha) to improve readability.
5. Explain what a larger point represents.
6. Identify whether the longest waits tend to occur on shorter or longer trips.
```

```python
distance = np.array([2, 3, 5, 7, 8, 10, 12, 15, 18, 20])
fare = np.array([9, 11, 15, 20, 22, 27, 31, 38, 45, 49])
wait = np.array([3, 5, 4, 6, 8, 7, 10, 12, 9, 15])

plt.scatter(
    distance,
    fare,
    s=wait * 15,
    alpha=0.6
)

plt.title("Distance vs Fare")
plt.xlabel("Distance (km)")
plt.ylabel("Fare ($)")

plt.grid()

plt.show()

# Larger points represent trips with longer waiting times.
```

```python
# Lab 6 — Histogram: Waiting-Time Distribution

Tasks:
1. Create a histogram of wait_times.
2. Use bins=6.
3. Add the title: Distribution of Passenger Waiting Time.
4. Label both axes.
5. Describe the overall shape of the distribution.
6. Is the distribution approximately symmetric, left-skewed, or right-skewed?
7. Which observations may be responsible for the longer tail?
```

```python
wait_times = np.array([
    3, 4, 5, 5, 6, 6, 6, 7, 7, 8,
    8, 9, 9, 10, 10, 11, 12, 13, 16, 22
])

plt.hist(wait_times, bins=6, edgecolor='black')

plt.title("Distribution of Passenger Waiting Time")
plt.xlabel("Waiting Time (minutes)")
plt.ylabel("Frequency")

plt.show()

# The distribution is somewhat right-skewed. Most waits are below about 13 minutes, while values such as 16 and 22 minutes create a longer right tail.
```

```python
# Lab 8 — Pie Chart: Mode Share
    
Tasks:
1. Create a pie chart using the data.
2. Add category labels.
3. Start the first slice at 90 degrees.
4. Display percentages on the chart.
5. Add the title: Transportation Mode Share.
6. Which mode has the largest share?
7. Which mode has the smallest share?
8. Would a bar chart make comparison between categories easier? Explain briefly.
```

```python
shares = np.array([40, 30, 20, 10])
labels = ["Metro", "Bus", "Taxi", "Bike"]

plt.pie(
    shares,
    labels=labels,
    startangle=90,
    autopct='%1.1f%%'
)

plt.title("Transportation Mode Share")

plt.show()    

# Metro has the largest share at 40%.

# Would a bar chart sometimes be better than a pie chart?
# Yes. Bar charts usually make differences between categories easier to compare.
```
