# With the while loop we can execute a set of statements as long as a condition is true.

```python
i = 1
while i < 6:   # (i<6) returns a boolean value (True/False) 
  print(i)
  i += 1
    
# Note: remember to increment i, or else the loop will continue forever.
```

```python
# With the break statement we can stop the loop even if the while condition is true:

i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```

```python
# With the continue statement we can stop the current iteration, and continue with the next:

i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```

# # List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
# newlist = []

# for x in fruits:
    
#   if "a" in x:
#     newlist.append(x)
    
    
    
newlist = [x for x in fruits if "a" in x]  # Python style

print(newlist)
```

```python
newlist = []
for x in fruits:
    if "a" in x:
        newlist.append(x)
        
print(newlist)
```

```python
# You can use the range() function to create an iterable:
newlist = [x for x in range(10)]   # left included, right excluded
newlist
```

```python
# Accept only even numbers
newlist = [x for x in range(10) if x % 2 == 0]
newlist
```

```python
# Set the values in the new list to upper case:
newlist = [x.upper() for x in fruits]
newlist
```

```python
newlist = ['hello' for x in fruits]
newlist
```

```python
newlist = [x if x != "banana" else "orange" for x in fruits]
newlist
```

# Remove Dictionary Items

```python
# The pop() method removes the item with the specified key name:

thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict.pop("model")
print(thisdict) 
```

```python
# The popitem() method removes the last inserted item (in versions before 3.7, a random item is removed instead):

thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict.popitem()
print(thisdict) 
```

```python
# The del keyword removes the item with the specified key name:

thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
del thisdict["model"]
print(thisdict) 
```

```python
# The del keyword can also delete the dictionary completely:

thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
del thisdict
print(thisdict) #this will cause an error because "thisdict" no longer exists. 
```

# Copy a Dictionary
## You cannot copy a dictionary simply by typing dict2 = dict1, because: dict2 will only be a reference to dict1, and changes made in dict1 will automatically also be made in dict2.

## There are ways to make a copy, one way is to use the built-in Dictionary method copy().

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
mydict = thisdict.copy()
print(mydict)
```

```python
# Make a copy of a dictionary with the dict() function:

thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
mydict = dict(thisdict)
print(mydict)
```
