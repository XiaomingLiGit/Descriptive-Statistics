# Part E2 — Functions

```python
# Q40: Waiting-time category

def wait_category(wait_min):
    if wait_min < 5:
        return "short"
    elif wait_min < 10:
        return "medium"
    else:
        return "long"


print(wait_category(3))
print(wait_category(7))
print(wait_category(15))
```

```python
# Q41: Fare per kilometre function

def fare_per_km(fare, distance):
    return fare / distance


print(fare_per_km(30, 10))
```

```python
# Q42: Long-wait function

def is_long_wait(wait_min):
    return wait_min > 10


print(is_long_wait(8))
print(is_long_wait(15))
```

```python
# Q43: Trip cost after tax

def trip_cost(base_fare, tax_rate):
    return base_fare * (1 + tax_rate)


print(trip_cost(25, 0.15))
```

```python
# Q44: Average speed

def speed_kmh(distance_km, duration_min):
    duration_hr = duration_min / 60
    return distance_km / duration_hr


print(speed_kmh(15, 30))
```

```python
# Q45: Protect against zero duration

def speed_kmh(distance_km, duration_min):

    if duration_min == 0:
        return "Invalid duration"

    duration_hr = duration_min / 60

    return distance_km / duration_hr


print(speed_kmh(15, 30))
print(speed_kmh(15, 0))
```

```python
# Q46: Apply a function to every waiting time

wait_times = [4.2, 8.5, 6.1, 12.3, 5.6, 15.2]

for x in wait_times:
    print(x, wait_category(x))
```
