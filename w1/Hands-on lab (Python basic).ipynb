```python
# Q1–Q4: Create variables and inspect their data types

trip_id = 101
distance_km = 8.5
zone = "Downtown"
completed = True

print(trip_id)
print(distance_km)
print(zone)
print(completed)

print(type(trip_id))
print(type(distance_km))
print(type(zone))
print(type(completed))
```

```python
# Q5: Convert a float to an integer

duration_min = 22.5

duration_integer = int(duration_min)

print(duration_min)
print(duration_integer)

# 22.5 becomes 22 — the decimal part is removed.
```

```python
# Q6: Convert a string to a numeric value

fare = "25.50"

print(type(fare))

fare = float(fare)

print(fare)
print(type(fare))
```

# Part B — Arithmetic and Comparisons

```python
distance_km = 12
duration_min = 30
fare = 28.50
wait_min = 8
```

```python
# Q7: Fare per kilometre

fare_per_km = fare / distance_km

print(fare_per_km)
```

```python
# Q8: Average speed in km/h

duration_hr = duration_min / 60

speed_kmh = distance_km / duration_hr

print(speed_kmh)
```

```python
# Q9: Increase fare by 15%

new_fare = fare * 1.15

print(new_fare)
```

```python
# Q10: Is the trip longer than 10 km?

print(distance_km > 10)
```

```python
# Q11: Is the waiting time at least 10 minutes?

print(wait_min >= 10)
```

```python
# Q12: Is fare between $20 and $40?

print((fare >= 20) and (fare <= 40))
```

```python
# Q13: Create a Boolean variable

long_trip = distance_km > 10

print(long_trip)
```

# Part C — Lists

```python
wait_times = [4.2, 8.5, 6.1, 12.3, 5.6, 9.8]
```

```python
# Q14: First waiting time

print(wait_times[0])
```

```python
# Q15: Last waiting time

print(wait_times[-1])
```

```python
# Q16: Number of observations

print(len(wait_times))
```

```python
# Q17: Total waiting time

print(sum(wait_times))
```

```python
# Q18: Average waiting time without NumPy or pandas

average_wait = sum(wait_times) / len(wait_times)

print(average_wait)
```

```python
# Q19: Minimum and maximum

print("Minimum:", min(wait_times))
print("Maximum:", max(wait_times))
```

```python
# Q20: Add another waiting time

wait_times.append(7.4)

print(wait_times)
```

```python
# Q21: Count waiting times > 8

count = 0

for x in wait_times:
    if x > 8:
        count += 1

print("Number of waits > 8:", count)
```
