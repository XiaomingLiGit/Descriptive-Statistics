# Lab 11 — Exploring Passenger Waiting Times with Violin Plots

Research question: How do waiting-time distributions differ across Bike, Bus, and Taxi services?

```python
# Part A — Create the starter dataset
# Task 1 — Generate and inspect the dataset


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(42)

# Bike: approximately symmetric
bike = rng.normal(4.5, 1.2, 60)

# Bus: right-skewed
bus = rng.lognormal(
    mean=np.log(7.5),
    sigma=0.55,
    size=60
)

# Taxi: two different waiting-time groups
taxi = np.concatenate([
    rng.normal(3.2, 0.8, 30),
    rng.normal(10.2, 1.2, 30)
])

df = pd.DataFrame({
    "service": ["Bike"]*60 + ["Bus"]*60 + ["Taxi"]*60,
    "wait_min": np.round(
        np.clip(
            np.concatenate([bike, bus, taxi]),
            0.5,
            None
        ),
        1
    )
})

display(df.head())
print(df.shape)
print(df["service"].value_counts())
```

```python
# Part B — Create a Violin Plot
# Task 2 — Compare waiting-time distributions

plt.figure(figsize=(9, 5))

sns.violinplot(
    data=df,
    x="service",
    y="wait_min",
    inner="box",
    cut=0
)

plt.title("Waiting Time Distribution by Service")
plt.xlabel("Transport Service")
plt.ylabel("Waiting Time (minutes)")

plt.show()
```

```python
# Part C — Violin Plot vs. Boxplot
# Task 3 — Create side-by-side comparisons

fig, axes = plt.subplots(
    1, 2,
    figsize=(13, 5)
)

sns.boxplot(
    data=df,
    x="service",
    y="wait_min",
    ax=axes[0]
)

axes[0].set_title("Boxplot")
axes[0].set_ylabel("Waiting Time (minutes)")

sns.violinplot(
    data=df,
    x="service",
    y="wait_min",
    inner="box",
    cut=0,
    ax=axes[1]
)

axes[1].set_title("Violin Plot")
axes[1].set_ylabel("Waiting Time (minutes)")

plt.tight_layout()
plt.show()
```

```python
# Part D — Validate the Visual Interpretation with Statistics
# Task 4 — Compute descriptive statistics

summary = df.groupby("service")["wait_min"].agg(
    count="count",
    mean="mean",
    median="median",
    std="std",
    minimum="min",
    maximum="max",
    skewness="skew"
)

display(summary.round(2))
```

```python
# Task 5 — Compute the IQR for each service

quartiles = df.groupby("service")[
    "wait_min"
].quantile([0.25, 0.75]).unstack()

quartiles.columns = ["Q1", "Q3"]

quartiles["IQR"] = (
    quartiles["Q3"] - quartiles["Q1"]
)

display(quartiles.round(2))
```

```python
# 1. Create a boxplot, using matplotlib

bike = df[df["service"] == "Bike"]["wait"]
bus = df[df["service"] == "Bus"]["wait"]
taxi = df[df["service"] == "Taxi"]["wait"]

data = [bike, bus, taxi]

plt.boxplot(
    data,
    labels=["Bike", "Bus", "Taxi"]
)

for i, values in enumerate(data, start=1):
    x = np.full(len(values), i)
    plt.scatter(x, values)

plt.title("Waiting Times Across Services")
plt.xlabel("Service")
plt.ylabel("Waiting Time (minutes)")

plt.show()
```

```python
# 2. Compare the median waiting times
df.groupby("service")["wait"].median()


# Bus has the highest median waiting time at 8.5 minutes.
# Taxi has the lowest median waiting time at 3.5 minutes.
```

```python
# 3. Compare the spread of the three groups
summary = df.groupby("service")["wait"].agg(
    Q1=lambda x: x.quantile(0.25),
    median="median",
    Q3=lambda x: x.quantile(0.75)
)

summary["IQR"] = summary["Q3"] - summary["Q1"]

summary


# Bus has the largest IQR (3.5 minutes), indicating greater variability in the middle 50% of waiting times.
# Bike and Taxi：IQR=2.5
# both of them with middle 50% spread 
```

4. Identify potential outliers






$Lower\ Fence=Q_1-1.5IQR$

$Upper\ Fence=Q_3+1.5IQR$

Bike:

$Q_1=3.75,\quad Q_3=6.25,\quad IQR=2.5$

Upper fence：6.25+1.5(2.5)=10
Since：15>10
Hence：15 minutes is a potential outlier for Bike.

Bus:

$Q_1=6.75,\quad Q_3=10.25,\quad IQR=3.5$

Upper fence：10.25+1.5(3.5)=15.5
Since：22>15.5
Hence：22 minutes is a potential outlier for Bus.

Taxi:

$Q_1=2.75,\quad Q_3=5.25,\quad IQR=2.5$

Upper fence：5.25+1.5(2.5)=9
Since：18>9
Hence：18 minutes is a potential outlier for Taxi.





| Service | Potential outlier |
|---|---:|
| Bike | 15 |
| Bus | 22 |
| Taxi | 18 |

Note: An outlier is not automatically an error.
