```python
import pandas as pd
import numpy as np
```

```python
data = {
 'trip_id': ['T001','T002','T003','T004','T005','T006','T007','T008','T009','T010'],
 'mode': ['Taxi','Bus','Bike','Taxi','Bus','Taxi','Bike','Bus','Taxi','Taxi'],
 'distance_km': [8.4,6.1,3.2,2.7,14.3,5.1,4.0,9.8,32.0,-1.5],
 'fare_cad': [19.5,3.75,0,11.2,3.75,13.2,0,3.75,72.0,15.0],
 'wait_min': [6,9,4,12,18,np.nan,3,11,55,8],
 'rating': [4,3,5,2,2,4,5,np.nan,1,3],
 'peak': ['No','Yes','No','Yes','Yes','No','No','Yes','Yes','Yes']
}

df = pd.DataFrame(data)
```

```python
# one-hot encoding for nominal categories, using Pandas
df_encoded = pd.get_dummies(df, columns=['mode'], dtype=int)

df_encoded
```

```python
# ordinal categorize the fare_level

levels = ['Low','Medium','High']


df['fare_level'] = pd.cut(
    df['fare_cad'],
    bins=[-np.inf, 5, 15, np.inf],
    labels=levels
)
```

```python
# (alternatively) , using lambda expression

df['fare_level'] = df['fare_cad'].apply(
    lambda x: 'Low' if x <= 5
    else 'Medium' if x <= 15
    else 'High'
)
```

```python
# ordinal encoding for ordinal categories, using Pandas

df['fare_encoded'] = df['fare_level'].cat.codes
```

```python
import pandas as pd
import torch
import torch.nn as nn

# Step 1: Convert mode categories to integer indices
mode_ids, categories = pd.factorize(df['mode'])

# Step 2: Create an embedding layer
embedding = nn.Embedding(
    num_embeddings=len(categories),
    embedding_dim=5
)

# Step 3: Generate embeddings for all rows
indices = torch.tensor(mode_ids, dtype=torch.long)

with torch.no_grad():
    vectors = embedding(indices).numpy()

# Step 4: Add embeddings to the original DataFrame
# df[['mode_emb_1', 'mode_emb_2']] = vectors

vectors
# display(df)
```
