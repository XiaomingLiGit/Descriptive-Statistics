```python
# Lab 1 — Build a probability table

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "period": ["Peak"] * 60 + ["Off-peak"] * 40,
    "status": (
        ["Late"] * 24
        + ["On time"] * 36
        + ["Late"] * 8
        + ["On time"] * 32
    )
})

# Frequency table
counts = pd.crosstab(
    df["period"],
    df["status"],
    margins=True
)

display(counts)

# Joint probability table
joint = pd.crosstab(
    df["period"],
    df["status"],
    normalize=True
)

display(joint)
```

```python
# Lab 2 — Complement and addition rules

p_late = (df["status"] == "Late").mean()

p_peak = (df["period"] == "Peak").mean()

p_peak_and_late = (
    (df["period"] == "Peak")
    & (df["status"] == "Late")
).mean()

p_not_late = 1 - p_late

p_peak_or_late = (
    p_peak + p_late - p_peak_and_late
)

print("P(Late):", p_late)
print("P(Not Late):", p_not_late)
print("P(Peak or Late):", p_peak_or_late)
```

```python
# Lab 3 — Conditional probability

# Conditional probability using Pandas
conditional = pd.crosstab(
    df["period"],
    df["status"],
    normalize="index"
)

display(conditional)

# Reverse the conditioning direction
reverse_conditional = pd.crosstab(
    df["status"],
    df["period"],
    normalize="index"
)

display(reverse_conditional)
```

```python
# Lab 4 — Test independence

p_independent = p_peak * p_late

print("Observed joint probability:", p_peak_and_late)
print("Product of marginal probabilities:", p_independent)

print(
    "Independent?",
    np.isclose(p_peak_and_late, p_independent)
)
```

```python
# Lab 5 — Expected value and simulation

import numpy as np

x = np.array([2, 5, 10])
p = np.array([0.20, 0.50, 0.30])

# Verify total probability
print("Total probability:", p.sum())

# Theoretical expected value
expected_value = np.sum(x * p)

print("E[X]:", expected_value)

# Simulate waiting times
rng = np.random.default_rng(42)

for n in [10, 100, 10000]:
    samples = rng.choice(
        x,
        size=n,
        p=p
    )

    print(
        f"n = {n:5d}, "
        f"Sample mean = {samples.mean():.3f}"
    )
```
